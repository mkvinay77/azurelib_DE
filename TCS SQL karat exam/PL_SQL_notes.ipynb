What is PL/SQL?
PL/SQL stands for Procedural Language extensions to SQL. It is Oracle's programming language that extends SQL by adding programming features such as variables, loops, conditions, and exception handling.

In simple words, SQL is used to work with data, while PL/SQL allows you to write a complete program that works with data.

## 1. SQL vs PL/SQL

| SQL                                                           | PL/SQL                                                                                    |
| ------------------------------------------------------------- | ----------------------------------------------------------------------------------------- |
| Used to query and manipulate data                             | Used to write programs that process data                                                  |
| Executes individual SQL statements                            | Executes a block of statements                                                            |
| Supports commands like `SELECT`, `INSERT`, `UPDATE`, `DELETE` | Supports SQL statements plus variables, loops, conditions, etc.                           |
| Example: Retrieve employee details                            | Example: Retrieve employees, check their salaries, and update records based on conditions |

## 2. Example of SQL

Suppose you have an `Employees` table.

```
SELECT emp_name, salary
FROM Employees
WHERE salary > 50000;
```

This SQL query retrieves employees whose salary is greater than ₹50,000.

## 3. Example of PL/SQL

Now suppose you want to retrieve an employee's salary and check whether it is greater than ₹50,000.

```
DECLARE
    v_salary NUMBER;
BEGIN
    SELECT salary
    INTO v_salary
    FROM Employees
    WHERE emp_id = 101;

    IF v_salary > 50000 THEN
        DBMS_OUTPUT.PUT_LINE('High Salary');
    ELSE
        DBMS_OUTPUT.PUT_LINE('Low Salary');
    END IF;
END;
/
```

Let's understand the important parts:

- `DECLARE` — declares variables.
- `BEGIN` — starts the executable section.
- `SELECT ... INTO` — retrieves the salary into a variable.
- `IF ... ELSE` — checks a condition.
- `DBMS_OUTPUT.PUT_LINE` — displays a message.
- `END;` — ends the PL/SQL block.
- `/` — tells Oracle tools such as SQL\*Plus to execute the completed block.

## 4. Structure of a PL/SQL block

DECLARE

Optional — variables and constants

BEGIN

Required — executable statements

EXCEPTION

Optional — handles runtime errors

END;

## 5. Where is PL/SQL used?

PL/SQL is primarily used in Oracle Database to:

- Create stored procedures and functions.
- Implement business logic inside the database.
- Process records using loops and conditions.
- Handle database errors.
- Create triggers that execute automatically when certain database events occur.

Interview tip: If you already know SQL, remember that PL/SQL adds procedural programming capabilities to SQL, allowing you to combine SQL statements with variables, conditions, loops, and exception handling.

One important point: PL/SQL is specific to Oracle, whereas Microsoft SQL Server uses T-SQL (Transact-SQL), its own extension of SQL.

Since you already practise SQL queries in SQL Server, the best way to learn PL/SQL is through hands-on practice with stored procedures and triggers, starting with basic syntax and progressing to interview-level examples.

One important difference before we start: SQL Server uses T-SQL, while Oracle Database uses PL/SQL. The concepts are similar, but the syntax is different.

## Lesson 1: Stored procedures in PL/SQL

A stored procedure is a named block of PL/SQL code stored in the Oracle database that you can execute whenever needed.

For example, instead of writing the same employee salary update query repeatedly, you can create a procedure once and call it whenever required.

### Step 1 — Create a practice table

Run this in your Oracle environment:

```
CREATE TABLE employees (
    emp_id NUMBER PRIMARY KEY,
    emp_name VARCHAR2(50),
    salary NUMBER(10, 2)
);

INSERT INTO employees VALUES (101, 'Rahul', 40000);
INSERT INTO employees VALUES (102, 'Priya', 60000);
INSERT INTO employees VALUES (103, 'Amit', 35000);

COMMIT;
```

### Step 2 — Create your first procedure

Suppose you want to increase an employee's salary.

```
CREATE OR REPLACE PROCEDURE increase_salary (
    p_emp_id IN NUMBER,
    p_amount IN NUMBER
)
AS
BEGIN
    UPDATE employees
    SET salary = salary + p_amount
    WHERE emp_id = p_emp_id;

    DBMS_OUTPUT.PUT_LINE('Salary updated successfully');
END;
/
```

Understand the syntax:

| Syntax                        | Meaning                                                      |
| ----------------------------- | ------------------------------------------------------------ |
| `CREATE OR REPLACE PROCEDURE` | Creates a procedure or replaces an existing one              |
| `increase_salary`             | Procedure name                                               |
| `p_emp_id IN NUMBER`          | Input employee ID                                            |
| `p_amount IN NUMBER`          | Input salary increment                                       |
| `AS`                          | Starts the procedure declaration section                     |
| `BEGIN ... END`               | Contains executable statements                               |
| `/`                           | Executes the completed definition in tools such as SQL\*Plus |

### Step 3 — Execute the procedure

```
EXEC increase_salary(101, 5000);
```

Expected message:

```
Salary updated successfully
```

Now verify the result:

```
SELECT *
FROM employees
WHERE emp_id = 101;
```

Expected salary: `45000`.

Notice that the procedure accepts different inputs, so you can reuse it for any employee.

Let's understand your PL/SQL procedure step by step, especially how the `FOR LOOP` works, how `emp` gets its values, and how the bonus is calculated and updated for each employee.

## 1. Complete code

```

CREATE OR REPLACE PROCEDURE CalculatePerformanceBonus
AS
    v_bonus NUMBER;
BEGIN
    FOR emp IN (
        SELECT EmployeeID, PerformanceScore
        FROM Employee
    )
    LOOP
        v_bonus := 100;

        IF emp.PerformanceScore >= 100 THEN
            v_bonus := v_bonus
                       + ((emp.PerformanceScore - 80) * 10)
                       + 200;

        ELSIF emp.PerformanceScore > 80 THEN
            v_bonus := v_bonus
                       + ((emp.PerformanceScore - 80) * 10);
        END IF;

        UPDATE Employee
        SET Bonus = v_bonus
        WHERE EmployeeID = emp.EmployeeID;

    END LOOP;

    COMMIT;
END;
/

```

## 2. Understand the `FOR LOOP` first

This is the most important part of your code.

```
FOR emp IN (
    SELECT EmployeeID, PerformanceScore
    FROM Employee
)
LOOP
```

Think of this as saying:

> Retrieve the employee ID and performance score for every employee, and process one employee at a time.

Suppose your `Employee` table contains these records:

| EmployeeID | PerformanceScore | Bonus |
| ---------- | ---------------- | ----- |
| 101        | 75               | NULL  |
| 102        | 85               | NULL  |
| 103        | 100              | NULL  |

The `SELECT` query retrieves these three rows.

The loop processes them one by one:

## 1

First iteration

`emp.EmployeeID = 101`

`emp.PerformanceScore = 75`

## 2

Second iteration

`emp.EmployeeID = 102`

`emp.PerformanceScore = 85`

## 3

Third iteration

`emp.EmployeeID = 103`

`emp.PerformanceScore = 100`

After processing the third employee, the loop ends because there are no more rows to process.

### What exactly is `emp`?

`emp` is a loop record variable that Oracle creates automatically for this cursor `FOR LOOP`.

You do not need to declare it separately.

Because your query selects two columns, you can access them through that record:

- `emp.EmployeeID`
- `emp.PerformanceScore`

You can think of `emp` as a temporary record holding the current employee's values during each iteration.

Important: `emp` is not the whole Employee table. It represents just the current row being processed.

## 3. What happens inside the loop?

### Step A: Initialize the bonus

```
v_bonus := 100;
```

Every time the loop starts a new iteration, `v_bonus` is reset to 100.

For example:

- Employee 101 starts with a bonus of 100.
- Employee 102 starts with a bonus of 100.
- Employee 103 starts with a bonus of 100.

This is important because each employee's bonus must be calculated independently.

`v_bonus` is a variable declared in the procedure, whereas `emp` is the loop record automatically created by Oracle.

### Step B: Check the first condition

```
IF emp.PerformanceScore >= 100 THEN
    v_bonus := v_bonus
               + ((emp.PerformanceScore - 80) * 10)
               + 200;
```

If the score is 100 or higher, the employee gets:

1. Base bonus of 100.
2. An additional 10 for every point above 80.
3. An extra 200 for scoring at least 100.

For an employee scoring 100:

\\[ 100 + ((100-80)\times10) + 200 = 500 \\]

So `v_bonus` becomes 500.

### Step C: Check `ELSIF`

```
ELSIF emp.PerformanceScore > 80 THEN
    v_bonus := v_bonus
               + ((emp.PerformanceScore - 80) * 10);
```

`ELSIF` means "else, if."

Oracle checks this condition only when the first `IF` condition is false.

For an employee scoring 85:

\\[ 100 + ((85-80)\times10) = 150 \\]

So `v_bonus` becomes 150.

What if the score is 75?

- Score greater than or equal to 100? No.
- Score greater than 80? No.
- The bonus remains 100.

### Step D: Update the employee's bonus

```
UPDATE Employee
SET Bonus = v_bonus
WHERE EmployeeID = emp.EmployeeID;
```

This writes the calculated bonus to the matching employee's row.

For example, during the first iteration:

```
UPDATE Employee
SET Bonus = 100
WHERE EmployeeID = 101;
```

During the second iteration, conceptually:

```
UPDATE Employee
SET Bonus = 150
WHERE EmployeeID = 102;
```

During the third iteration:

```
UPDATE Employee
SET Bonus = 500
WHERE EmployeeID = 103;
```

You don't need to write these three statements yourself. The loop executes the same `UPDATE` statement with different values each time.

Why do we use `WHERE EmployeeID = emp.EmployeeID`?

Without the `WHERE` condition, the `UPDATE` could change the bonus for every employee in the table during each iteration. The condition ensures that only the current employee's row is updated.

## 4. How does the entire procedure execute?

Start procedure

Fetch the next employee using `FOR emp IN (...)`

Set `v_bonus := 100`

Check performance score using `IF / ELSIF`

Update the current employee's bonus

Repeat for the next employee until no rows remain.

`COMMIT` — save the changes

End procedure

## 5. What do `COMMIT`, `END LOOP`, and `END` mean?

```
END LOOP;

COMMIT;
END;
/
```

- `END LOOP;` — finishes the loop body and moves to the next row. When there are no more rows, execution continues after the loop.
- `COMMIT;` — saves the transaction's changes to the database.
- `END;` — finishes the PL/SQL procedure's executable block.
- `/` — in Oracle tools that support script execution, submits the completed procedure definition for compilation.

Notice that `COMMIT` is outside the loop. Therefore, the procedure commits after it finishes processing all employees, rather than committing once per employee.
