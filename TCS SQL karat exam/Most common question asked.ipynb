1) Execution Plan for SQL in PL/SQL Application
What is an Execution Plan?

When Oracle receives a SQL query, it decides:

How should I fetch the data?


For example:

SELECT *
FROM employees
WHERE emp_id = 101;


Oracle can:

Use Index Scan
Use Full Table Scan
Use Range Scan

The path chosen by Oracle is called the Execution Plan.

Real-Life Example

Imagine finding a book.

Method 1
Go page by page


This is like:

Full Table Scan

Method 2
Use the book index


This is like:

Index Scan


Oracle chooses the faster method.

Check Execution Plan
EXPLAIN PLAN FOR
SELECT *
FROM employees
WHERE emp_id = 101;

SELECT * FROM TABLE(DBMS_XPLAN.DISPLAY);

Interview Answer

An execution plan is the sequence of operations Oracle uses to execute a SQL statement. It shows whether Oracle uses indexes, joins, sorting, or full table scans to retrieve data.

2) Nested Tables and Rules
What is a Nested Table?

A nested table is a collection datatype that can store multiple values in a single variable.

Think:

Variable = One value

Nested Table = Multiple values

Example
Create Type
CREATE TYPE num_list AS TABLE OF NUMBER;
/

Use It
DECLARE
   v_numbers num_list;
BEGIN
   v_numbers := num_list(10,20,30);

   FOR i IN 1..v_numbers.COUNT LOOP
      DBMS_OUTPUT.PUT_LINE(v_numbers(i));
   END LOOP;
END;
/


Output:

10
20
30

Rules
Rule 1

Must create collection type.

CREATE TYPE num_list AS TABLE OF NUMBER;

Rule 2

Elements can grow dynamically.

v_numbers.EXTEND;

Rule 3

Count elements using:

v_numbers.COUNT

Rule 4

Can store records, not only numbers.

TABLE OF VARCHAR2
TABLE OF NUMBER
TABLE OF OBJECTS

Interview Answer

Nested tables are collection datatypes that can store multiple rows of data in a single PL/SQL variable. They are dynamically sizable and can be used inside PL/SQL programs and database tables.

3) RETURN Clause in Function
Why Return?

A Function must return a value.

Example
CREATE OR REPLACE FUNCTION get_bonus
(
   p_salary NUMBER
)
RETURN NUMBER
AS
BEGIN
   RETURN p_salary * 0.10;
END;
/


Here:

RETURN NUMBER


means:

This function will return a NUMBER value.

Calling Function
SELECT get_bonus(50000)
FROM dual;


Output:

5000

Rule

Every function must have:

RETURN datatype


and

RETURN value;


inside its body.

Interview Answer

The RETURN clause defines the datatype returned by a function. Every PL/SQL function must return exactly one value whose datatype matches the RETURN datatype.

4) Operators and Rules

Operators perform calculations and comparisons.

Arithmetic Operators
Operator	Meaning+	Addition
-	Subtraction
*	Multiplication
/	Division

Example:

DECLARE
   v_result NUMBER;
BEGIN
   v_result := 10 + 5;

   DBMS_OUTPUT.PUT_LINE(v_result);
END;
/


Output:

15

Comparison Operators
Operator	Meaning=	Equal
<>	Not Equal
>	Greater Than
<	Less Than
>=	Greater Than Equal
<=	Less Than Equal

Example:

IF salary > 50000 THEN
   DBMS_OUTPUT.PUT_LINE('High Salary');
END IF;

Logical Operators
Operator	MeaningAND	Both true
OR	One true
NOT	Opposite

Example:

IF salary > 50000
AND bonus > 1000
THEN
   DBMS_OUTPUT.PUT_LINE('Eligible');
END IF;

String Operator
||


Concatenation

Example:

DECLARE
   v_name VARCHAR2(20);
BEGIN
   v_name := 'Vinay' || ' Kumar';

   DBMS_OUTPUT.PUT_LINE(v_name);
END;
/


Output:

Vinay Kumar

Interview Answer

PL/SQL supports arithmetic, comparison, logical, and string operators. They are used in calculations, conditions, and data manipulation.

5) How Many Ways Can We Validate Parameters in PL/SQL?

This is a favorite interview question.

There are multiple ways.

Method 1: IF Condition
CREATE OR REPLACE PROCEDURE add_emp
(
   p_salary NUMBER
)
AS
BEGIN

   IF p_salary < 0 THEN
      RAISE_APPLICATION_ERROR(
      -20001,
      'Salary cannot be negative');
   END IF;

END;
/

Method 2: NOT NULL Check
IF p_salary IS NULL THEN
   RAISE_APPLICATION_ERROR(
   -20002,
   'Salary cannot be null');
END IF;

Method 3: Range Validation
IF p_age NOT BETWEEN 18 AND 60 THEN
   RAISE_APPLICATION_ERROR(
   -20003,
   'Invalid age');
END IF;

Method 4: REGEXP Validation

Validate email.

IF NOT REGEXP_LIKE(
      p_email,
      '^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+$')
THEN
   RAISE_APPLICATION_ERROR(
   -20004,
   'Invalid Email');
END IF;

Method 5: Database Constraint Validation

Table:

CREATE TABLE employee(
  emp_id NUMBER PRIMARY KEY,
  salary NUMBER CHECK(salary > 0)
);


Oracle itself validates.

INSERT INTO employee
VALUES(1,-1000);


Result:

ERROR

Interview Answer

Parameters in PL/SQL can be validated using IF conditions, NULL checks, range checks, REGEXP validation, custom exceptions using RAISE_APPLICATION_ERROR, and database constraints like PRIMARY KEY, UNIQUE, CHECK, and FOREIGN KEY.
