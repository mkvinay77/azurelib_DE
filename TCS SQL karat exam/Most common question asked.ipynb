1) Execution Plan for SQL in PL/SQL Application
What is an Execution Plan?

When Oracle receives a SQL query, it decides:

How should I fetch the data?
For example:

SELECT *
FROM employees
WHERE emp_id = 101;


Oracle can:
Use Index Scan
Use Full Table Scan
Use Range Scan

The path chosen by Oracle is called the Execution Plan.

Real-Life Example

Imagine finding a book.

Method 1
Go page by page


This is like:

Full Table Scan

Method 2
Use the book index


This is like:

Index Scan


Oracle chooses the faster method.

Check Execution Plan
EXPLAIN PLAN FOR
SELECT *
FROM employees
WHERE emp_id = 101;

SELECT * FROM TABLE(DBMS_XPLAN.DISPLAY);

Interview Answer

An execution plan is the sequence of operations Oracle uses to execute a SQL statement. It shows whether Oracle uses indexes, joins, sorting, or full table scans to retrieve data.

2) Nested Tables and Rules
What is a Nested Table?

A nested table is a PL/SQL collection type that stores multiple elements of the same data type in a single collection variable. It can be used to process multiple values or rows together.

Think:

Variable = One value

Nested Table = Multiple values

Example
-- Create Type

/

   v_numbers num_list;

      DBMS_OUTPUT.PUT_LINE(v_numbers(i));
   END LOOP;
END;
/

30

Rule 1
Must create collection type.

CREATE TYPE num_list AS TABLE OF NUMBER;

Elements can grow dynamically.

Count elements using:

Rule 4
Can store records, not only numbers.

TABLE OF VARCHAR2
TABLE OF NUMBER
TABLE OF OBJECTS

DECLARE
    TYPE salary_type IS TABLE OF NUMBER
        INDEX BY PLS_INTEGER;

    v_salary salary_type;
BEGIN
    v_salary(1) := 30000;
    v_salary(2) := 40000;
    v_salary(3) := 50000;

    DBMS_OUTPUT.PUT_LINE(v_salary(2));
END;
/
```

Output:

```
40000
```

Explanation:

- `TYPE salary_type IS TABLE OF NUMBER` defines a collection of numbers.
- `INDEX BY PLS_INTEGER` makes it an associative array indexed by integers.
- `v_salary(2) := 40000` stores `40000` under key `2`.
- `v_salary(2)` retrieves the value stored under that key.

You don't have to populate every index sequentially. For example, you can use indexes `1`, `5`, and `10`.

## 2. Nested Table

A nested table stores multiple elements of the same type. Its size can grow dynamically, and elements can be deleted.

Think of it as a Python list.

### Example

```
DECLARE
    TYPE employee_ids_type IS TABLE OF NUMBER;

    v_employee_ids employee_ids_type;
BEGIN
    v_employee_ids := employee_ids_type(101, 102, 103);

    DBMS_OUTPUT.PUT_LINE(v_employee_ids(1));
    DBMS_OUTPUT.PUT_LINE(v_employee_ids(2));
    DBMS_OUTPUT.PUT_LINE(v_employee_ids(3));
END;
/
```

Output:

```
101
102
103
```

Explanation:

- `TYPE employee_ids_type IS TABLE OF NUMBER` defines a nested table collection type.
- `employee_ids_type(101, 102, 103)` initializes the collection with three elements.
- `v_employee_ids(1)` retrieves the first element.

Important: A nested table variable must be initialized before you can use it this way. Nested tables can also be defined at the SQL level and stored in database tables.

## 3. VARRAY

VARRAY stands for Variable-Size Array. It stores an ordered collection of elements, but you must define its maximum size when creating the type.

### Example

```
DECLARE
    TYPE top_scores_type IS VARRAY(3) OF NUMBER;

    v_scores top_scores_type;
BEGIN
    v_scores := top_scores_type(80, 90, 95);

    DBMS_OUTPUT.PUT_LINE(v_scores(1));
    DBMS_OUTPUT.PUT_LINE(v_scores(2));
    DBMS_OUTPUT.PUT_LINE(v_scores(3));
END;
/
```

Output:

```
80
90
95
```

Explanation:

- `VARRAY(3)` means the collection can contain at most three elements.
- `top_scores_type(80, 90, 95)` initializes it with three values.
- The elements retain their order.

You cannot add a fourth element while the collection already contains three elements.

## 4. Difference between the three collection types

| Feature                            | Associative Array                        | Nested Table                                      | VARRAY                                                   |
| ---------------------------------- | ---------------------------------------- | ------------------------------------------------- | -------------------------------------------------------- |
| Indexing                           | Integer or string keys                   | Integer indexes                                   | Integer indexes                                          |
| Maximum size                       | No predefined limit                      | No predefined limit                               | Fixed maximum                                            |
| Can have gaps in indexes?          | Yes                                      | Yes, after deleting elements                      | Not in the same way; elements form an ordered collection |
| Initialization                     | Not required before assigning an element | Required before using an uninitialized collection | Required before using an uninitialized collection        |
| Can be stored in a database table? | Generally no, as a SQL column collection | Yes, with SQL-level type definitions              | Yes, with SQL-level type definitions                     |
| Common use                         | Lookup data by key                       | Flexible collections of elements                  | Small, ordered collections with a maximum size           |

Note: Nested tables and VARRAYs can be used in both PL/SQL and SQL when defined appropriately. Associative arrays are primarily PL/SQL collections, though Oracle supports some associative-array use in SQL contexts.

## 5. Real-world example

Suppose you are working as a Data Engineer and need to process employee salaries.

Associative Array

Store employee salaries using employee IDs as keys.

Example: employee ID 101 → salary 50,000

Nested Table

Store a flexible list of employee IDs retrieved from a query.

Example: 101, 102, 103, 104, …

VARRAY

Store a small, ordered list with a known maximum size.

Example: the top 3 salary values

## 6. Interview-ready answer

If the interviewer asks, “What are collection types in PL/SQL?”, you can say:

> Oracle PL/SQL provides three main collection types: associative arrays, nested tables, and VARRAYs. Associative arrays store elements using keys and are useful for lookups. 
Nested tables are flexible collections that can grow dynamically and allow elements to be deleted. 
VARRAYs store ordered elements with a predefined maximum size. The appropriate type depends on how the data needs to be accessed and managed.

Remember this shortcut:

- Associative array = Key-value lookup
- Nested table = Flexible collection
- VARRAY = Ordered collection with a maximum size


1. SELECT INTO in PL/SQL
What is SELECT INTO?
SELECT INTO is used in PL/SQL to retrieve a single row from a query and store the result into one or more variables.
Important Rule:::
SELECT INTO must return exactly one row.
Example

Employee table:
EMP_ID EMP_NAME SALARY
101    VINAY    50000

Code:
SQL
DECLARE
v_salary NUMBER;
BEGIN
SELECT salary
INTO v_salary
FROM employee
WHERE emp_id = 101;
DBMS_OUTPUT.PUT_LINE(v_salary);
END;
/

2. Declaring and Initializing Collections
What is a Collection?
Collections are PL/SQL data structures that store multiple values. Common collection types are Nested Tables, VARRAYs, and Associative Arrays.
Methods
1 COUNT
v_numbers.COUNT
Number of elements.
2 EXTEND
v_numbers.EXTEND;
Adds a new element.
3 DELETE
v_numbers.DELETE;
Removes elements.

3. Function Cache and Conditions?
What is Function Result Cache?
Instead of executing the same function repeatedly, Oracle stores the result in memory.
Next time same input comes, Oracle returns the cached result.

Function RESULT_CACHE stores function results in memory. When the same input parameters are supplied again, Oracle returns the cached result instead of re-executing the function.

4. SAVEPOINT
What is SAVEPOINT?
A SAVEPOINT creates a checkpoint inside a transaction.
You can rollback partially to that point instead of rolling back everything.

or
SAVEPOINT is used to mark a checkpoint within a transaction. It allows partial rollback to a specific point instead of rolling back the entire transaction.

Example:::
BEGIN

    INSERT INTO employee VALUES(1,'A');

    SAVEPOINT sp1;

    INSERT INTO employee VALUES(2,'B');

    SAVEPOINT sp2;

    INSERT INTO employee VALUES(3,'C');
END;
/

Current Data:
Plain Text
1 A
2 B
3 C

ROLLBACK TO sp2;
Result:
Plain Text
1 A
2 B
Row C removed.

5. If Named PL/SQL Subprograms Get Overloaded
What is Overloading?
Overloading means creating multiple procedures or functions with the same name but different parameter lists. Oracle resolves the correct subprogram based on the number, datatype, or mode of parameters.
Rules for Overloading

✅ Different number of parameters

SQL
func(p_num)
func(p_num1,p_num2)

✅ Different parameter types

SQL
func(NUMBER)
func(VARCHAR2)

❌ Return type alone cannot overload

Invalid:

SQL
FUNCTION f(p NUMBER)
RETURN NUMBER;
FUNCTION f(p NUMBER)
RETURN VARCHAR2;

Oracle cannot determine which one to call.
