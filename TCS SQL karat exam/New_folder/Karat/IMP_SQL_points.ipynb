1. LEFT SEMI JOIN Question
Question

Find all books that have been borrowed by at least one borrower.

Solution using EXISTS
SELECT *
FROM Books b
WHERE EXISTS (
    SELECT 1
    FROM Borrowers br
    WHERE br.BorrowedBookId = b.BookId
);

Output
101  R.K. Narayan
102  George Orwell
103  J.K. Rowling
105  Chetan Bhagat

Explanation

For Book 101:

Borrower exists ? YES


For Book 104:

Borrower exists ? NO


So Book 104 is excluded.

✅ This behaves like a LEFT SEMI JOIN.

2. LEFT ANTI JOIN Question
Question

Find all books that have never been borrowed.

Solution using NOT EXISTS
SELECT *
FROM Books b
WHERE NOT EXISTS (
    SELECT 1
    FROM Borrowers br
    WHERE br.BorrowedBookId = b.BookId
);

Output
104  Dan Brown

Explanation
101 -> Borrowed
102 -> Borrowed
103 -> Borrowed
104 -> Not Borrowed
105 -> Borrowed


✅ This behaves like a LEFT ANTI JOIN.

Interview Shortcut

LEFT SEMI JOIN

Return rows from Left Table if a match exists in Right Table.

WHERE EXISTS (...)


Example:

Find books that have been borrowed.


LEFT ANTI JOIN

Return rows from Left Table if NO match exists in Right Table.

WHERE NOT EXISTS (...)


Example:

Find books that have never been borrowed.

Easy Memory Trick
EXISTS      = Match Found      = SEMI JOIN

NOT EXISTS  = Match Not Found  = ANTI JOIN


These are the most common interview patterns for SEMI JOIN and ANTI JOIN questions.

Almost ✅, but not exactly.

Conceptually
EXISTS  ≈ IN

NOT EXISTS ≈ NOT IN


Both pairs are often used to solve the same business problem.

⚠️ Problem:

Subquery returns:

101
102
103
105
NULL


because Karthik has:

BorrowedBookId = NULL


Now Oracle evaluates:

Output:

No Rows Returned


❌ Wrong result.

Correct NOT IN
SELECT *
FROM Books
WHERE BookId NOT IN (
    SELECT BorrowedBookId
    FROM Borrowers
    WHERE BorrowedBookId IS NOT NULL
);


Now output:

✅ Correct.

Interview Answer
EXISTS and IN are often interchangeable.

NOT EXISTS and NOT IN are often interchangeable.

However, NOT EXISTS is safer because it handles NULL values correctly, whereas NOT IN can return unexpected results if the subquery contains NULL.

Rule of Thumb
SEMI JOIN  -> EXISTS / IN

ANTI JOIN  -> NOT EXISTS / NOT IN

Preferred in Interviews
EXISTS
NOT EXISTS


because they avoid NULL-related issues and clearly express the intent. ✅

Q) Find all books that have been borrowed by at least one borrower.
Yes ✅, INNER JOIN can solve this question too.

The difference is: When to use INNER JOIN V/S EXSIST

**EXISTS returns the book once if a match exists.
INNER JOIN returns one row for each matching combination.**

Let's see with sample data.

Scenario 1: One borrower per book
Books
BookId  Author
------  ---------------
101     R.K. Narayan
102     George Orwell
103     J.K. Rowling
104     Dan Brown
105     Chetan Bhagat

Borrowers
BorrowerId  FirstName  BorrowedBookId
----------  ---------  --------------
1           Arun       101
2           Priya      103
3           Rahul      102
4           Divya      105

INNER JOIN
SELECT b.BookId,
       b.Author
FROM Books b
INNER JOIN Borrowers br
ON b.BookId = br.BorrowedBookId;


Output:

101  R.K. Narayan
102  George Orwell
103  J.K. Rowling
105  Chetan Bhagat

EXISTS
SELECT *
FROM Books b
WHERE EXISTS (
    SELECT 1
    FROM Borrowers br
    WHERE br.BorrowedBookId = b.BookId
);


Output:

101  R.K. Narayan
102  George Orwell
103  J.K. Rowling
105  Chetan Bhagat


✅ Same result.
