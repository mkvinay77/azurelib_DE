```python
"""
Q) i have a below csv which contain the following column like 
emp deatiles id,name,date,salary


now i want to write a pyspark code in such way that the result should emp salary in quater wise like id, name, q1,q2,q3,q4 like this 




from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.appName("QuarterSalary").getOrCreate()

df = spark.read.csv(
    "/path/emp_details.csv",
    header=True,
    inferSchema=True
)

result_df = (
    df.withColumn("date", to_date(col("date"), "yyyy-MM-dd"))
      .withColumn("quarter", concat(lit("Q"), quarter(col("date"))))
      .groupBy("id", "name")
      .pivot("quarter", ["Q1", "Q2", "Q3", "Q4"])
      .sum("salary")
)

result_df.show()

"""
```

```python
data = [
    (1, "Vinay", "2026-01-15", 1000),
    (1, "Vinay", "2026-04-10", 2000),
    (1, "Vinay", "2026-08-20", 3000),
    (2, "Raj",   "2026-02-05", 1500),
    (2, "Raj",   "2026-11-01", 2500)
]
from pyspark.sql.functions import *

cols = ["id", "name", "date", "salary"]

df = spark.createDataFrame(data, cols)

df1 = (
        df.withColumn("date", to_date(col("date"), "yyyy-MM-dd")) 
          .withColumn("quarter", concat(lit("Q"), quarter(col("date"))))
       )
display(df1)

result_df = (
              df1.groupBy(col("id"),col("name"))
                 .pivot("quarter",["Q1","Q2","Q3","Q4"])
                 .agg(sum(col("salary")))
)
display(result_df)


"""
Quarter Mapping
Plain Text
Jan Feb Mar -> Q1
Apr May Jun -> Q2
Jul Aug Sep -> Q3
Oct Nov Dec -> Q4

Examples:

Plain Text
2026-01-10 -> 1
2026-05-10 -> 2
2026-08-10 -> 3
2026-11-10 -> 4

6. concat()
Syntax
Python
concat(column1, column2)
Purpose

Joins multiple values together.

Example
Python
concat(lit("Q"), quarter(col("date")))

Suppose:

Plain Text
quarter(date) = 2

Then:

Plain Text
"Q" + "2"

Output:

Plain Text
Q2

Examples:

Plain Text
Q + 1 = Q1
Q + 2 = Q2
Q + 3 = Q3
Q + 4 = Q4

pivot()
Syntax
Python
pivot("column")
Purpose

Converts row values into columns.

Before Pivot
Plain Text
id name quarter salary
1 Vinay Q1 1000
1 Vinay Q2 2000
1 Vinay Q3 3000
After Pivot

-- Plain Text
id name Q1 Q2 Q3
1 Vinay 1000 2000 3000
"""
```
