What is Agile?

Agile is a software development approach where work is delivered in small, iterative cycles (Sprints) instead of waiting months for one big release.

Traditional (Waterfall)
Requirement
    ↓
Development
    ↓
Testing
    ↓
Deployment


Time:

4-6 Months


Problem:

Business waits a long time to see results.

Agile
Requirement
    ↓
Sprint 1 → Release
    ↓
Sprint 2 → Release
    ↓
Sprint 3 → Release


Benefit:

Faster delivery
Continuous feedback
Early issue detection
Better quality

Interview Answer (2-3 Minutes)

"We follow Agile Scrum methodology with 2-week sprints. Business requirements are captured in Jira and maintained in the product backlog. Large requirements are created as Epics and broken into user stories. During backlog refinement and sprint planning, stories are estimated using story points and selected for the sprint. As a Data Engineer, I work on ADF pipelines, PySpark transformations, and data validation tasks. We attend daily standups to discuss progress and blockers. After development, code goes through code review, testing, and UAT. Once approved, it is deployed to production. At the end of the sprint, we conduct sprint review and retrospective meetings before starting the next sprint."

This answer is commonly sufficient for "Explain Agile and how it works in your project" interview questions.
