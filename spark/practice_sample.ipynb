```python
data_A = [
    ("swd", "wsw", "qiak", 1234),
    ("abc", "xyz", "rst", 456),
    ("rashmi", "op", "qr", 487),
    ("wrt", "rqt", "my", 567),
    ("yu", "igh", "thg", 987),
    ("fth", "srg", "pio", 323)
]

columns_A = ["first_name", "middle_name", "last_name", "empid"]

df1 = spark.createDataFrame(data_A, columns_A)
```

```python
data_B = [
    (567, 1000),
    (456, 2000),
    (987, 5000),
    (567, 8000),
    (987, 500),
    (764, 5000)
]

columns_B = ["empid", "salary"]

df2 = spark.createDataFrame(data_B, columns_B)
```

```python
#Join 
join_df = df1.join(df2,"empid", "inner")
display(join_df)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *

window_spec = Window.orderBy(col("salary").desc())

final_df = join_df.withColumn("rnk", dense_rank().over(window_spec))\
                #    .filter(col("rnk") == 3)

display(final_df)
```

```python
final_df.write \
    .mode("overwrite") \
    .format("csv") \
    .option("header", "true") \
    .save("/Volumes/workspace/default/batch12-volume/final_rnk")
```
