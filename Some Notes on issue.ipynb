No, Scrum Master and ITPM are usually different roles, though responsibilities can overlap depending on the company.

1. Scrum Master

In Agile projects, the Scrum Master's responsibility is to:

Conduct Daily Standups
Sprint Planning
Sprint Review
Sprint Retrospective
Remove team impediments/blockers
Ensure Agile process is followed

Example:

Your ADF pipeline is blocked because another team has not provided source access. You raise it in standup, and the Scrum Master helps coordinate and remove the blocker.

2. ITPM (IT Project Manager)

ITPM = Information Technology Project Manager

Responsibilities:

Overall project delivery
Timeline tracking
Resource planning
Risk management
Stakeholder communication
Budget and project governance
Escalation management

Example:

If a source system issue is delaying your sprint and may impact the release date, the ITPM may coordinate with multiple teams, managers, and stakeholders to ensure the issue gets resolved.

This is a very common behavioral/situational interview question.

Question:

"Suppose you are stuck with a task or facing a blocker. What will you do?"

Bad Answer ❌

"I will directly escalate to my manager or ITPM."

Interviewers may think:

You don't try to solve problems on your own.
You escalate too quickly.
Good Answer ✅

"First, I will spend some time analyzing the issue myself by checking logs, error messages, previous implementations, documentation, and relevant tickets. If I am still unable to resolve it, I will discuss it with team members or senior colleagues to see if they have encountered a similar issue before. If the blocker is impacting sprint deliverables or timelines, I will raise it during the daily stand-up and communicate the risk to the Scrum Master/ITPM and my manager. I will clearly explain what has been tried, the impact, and any support needed. This ensures transparency and helps avoid delays in sprint commitments."

If They Ask Specifically About Sprint Calls

You can answer:

"During the daily sprint stand-up, I would mention the blocker, explain the impact on my story, and share the troubleshooting steps I have already performed. If additional support is needed from another team or stakeholder, I would request help through the Scrum Master or ITPM so that the dependency can be tracked and resolved quickly."

Data Engineer Example

"Suppose an ADF pipeline is failing because the source database team has changed a table structure. First, I would validate the error logs, confirm the root cause, and check whether any quick fix is available. If the dependency is external and cannot be resolved by me immediately, I would update the blocker in Jira and communicate it during the stand-up. If it threatens the sprint timeline, I would escalate it to the ITPM and manager with details of the impact and expected support required."

Short Interview-Friendly Answer

"I first try to troubleshoot independently. If I'm still blocked, I seek help from teammates or SMEs. If the issue impacts sprint delivery, I communicate it in the stand-up, update Jira, and escalate to the Scrum Master/ITPM or manager with complete details of the blocker and its impact."

This answer shows:

✅ Problem-solving

✅ Agile working style

Interview Question

"What will you do if you're stuck on a task?"

Answer with Real Project Example

In my current project, we use Agile methodology and track our work through Jira. If I get stuck on a task, I first try to analyze the issue myself by checking logs, code, pipeline runs, and previous implementations. I usually spend some reasonable time troubleshooting before asking for help.

For example, I was working on an Azure Data Factory incremental pipeline that was loading data from SQL Server to ADLS. Suddenly, the pipeline started failing because a new column had been added in the source table, causing schema mismatch issues in the downstream PySpark job.

My approach was:

Checked the ADF pipeline error logs.
Validated the source table schema.
Compared the current schema with the expected schema in the PySpark transformation layer.
Tried implementing a fix on my own.

Since the issue required coordination with the source system team, I raised it during the daily stand-up and updated the Jira ticket with blocker details.

I informed the Scrum Master/ITPM that the dependency was external and could impact the sprint timeline if not resolved quickly.

The ITPM coordinated with the source team, the schema details were confirmed, and we modified the pipeline accordingly. The issue was resolved without affecting the production release.

This approach ensured:

Ownership from my side
Proper communication
Visibility to stakeholders
Timely escalation of risks
Short 1-Minute Version

"If I'm stuck, I first troubleshoot independently by checking logs, code, and documentation. If I'm unable to resolve the issue, I discuss it with team members or SMEs. If the blocker can impact sprint commitments, I communicate it during the stand-up, update Jira, and inform the Scrum Master/ITPM. For example, in one project an ADF incremental pipeline failed due to a source schema change. After validating the issue, I raised it as a blocker during the sprint call, coordinated with the source team through the ITPM, and the issue was resolved before the release deadline."
