What is a Trigger?

A Trigger is a PL/SQL block that executes automatically when a specific event occurs on a table.

Events are usually:

INSERT
UPDATE
DELETE


Think:

Event occurs
      ↓
Trigger fires automatically

Real Life Example

When you open a door:

Door Opened
     ↓
Light Turns ON Automatically


You didn't manually switch on the light.

Similarly:

Employee Inserted
       ↓
Trigger Executes Automatically

Why Do We Use Triggers?

Triggers are mainly used for:

1. Auditing

Track who changed what and when.

Example:

Salary changed
↓
Store old salary and new salary in audit table

2. Data Validation

Prevent invalid data.

Example:

Salary < 0
↓
Reject insert

3. Automatic Logging
Employee Added
↓
Store activity in log table

4. Business Rules Enforcement

Example:

Employee cannot be deleted after retirement.


Trigger can enforce this rule.

Trigger Syntax
CREATE OR REPLACE TRIGGER trigger_name
BEFORE INSERT
ON employee
FOR EACH ROW
BEGIN

   -- logic

END;
/

Example 1: BEFORE INSERT Trigger

Suppose salary should never be negative.

CREATE OR REPLACE TRIGGER trg_check_salary
BEFORE INSERT
ON employee
FOR EACH ROW
BEGIN

   IF :NEW.salary < 0 THEN
      RAISE_APPLICATION_ERROR(
      -20001,
      'Salary cannot be negative');
   END IF;

END;
/


Now if user runs:

INSERT INTO employee
VALUES(101,'VINAY',-5000);


Result:

ERROR
Salary cannot be negative


Trigger stopped bad data.

Example 2: AFTER INSERT Trigger

Create Audit Table:

CREATE TABLE emp_audit
(
   emp_id NUMBER,
   action_type VARCHAR2(20),
   action_date DATE
);


Trigger:

CREATE OR REPLACE TRIGGER trg_emp_audit
AFTER INSERT
ON employee
FOR EACH ROW
BEGIN

   INSERT INTO emp_audit
   VALUES
   (
      :NEW.emp_id,
      'INSERT',
      SYSDATE
   );

END;
/


When somebody inserts:

INSERT INTO employee
VALUES(101,'VINAY',50000);


Oracle automatically inserts:

101 INSERT 21-OCT-2026


into audit table.

Understanding :NEW and :OLD

This is extremely important.

Suppose data:

EMP_ID  SALARY

101     50000


Update:

UPDATE employee
SET salary = 70000
WHERE emp_id = 101;


Before update:

50000


After update:

70000


Trigger sees:

:OLD.salary = 50000
:NEW.salary = 70000

Salary Audit Example
CREATE OR REPLACE TRIGGER trg_salary_change
AFTER UPDATE
ON employee
FOR EACH ROW
BEGIN

   INSERT INTO salary_audit
   (
      emp_id,
      old_salary,
      new_salary
   )
   VALUES
   (
      :OLD.emp_id,
      :OLD.salary,
      :NEW.salary
   );

END;
/

Types of Triggers
1. BEFORE Trigger

Runs before operation.

BEFORE INSERT
BEFORE UPDATE
BEFORE DELETE


Used for:

Validation
Modification
Business Rules


Example:

Check salary
Check age
Check mandatory fields

2. AFTER Trigger

Runs after operation.

AFTER INSERT
AFTER UPDATE
AFTER DELETE


Used for:

Audit
Logging
Tracking
Notifications

Row-Level Trigger
FOR EACH ROW


Example:

AFTER INSERT
ON employee
FOR EACH ROW


Suppose:

INSERT 100 rows


Trigger executes:

100 times


One for each row.

Statement-Level Trigger

Without:

FOR EACH ROW


Example:

AFTER INSERT ON employee


Insert:

100 rows


Trigger executes:

1 time only

Procedure vs Trigger

This is asked very frequently.

Procedure
EXEC update_salary(101,10);


You call it manually.

Manual execution

Trigger
INSERT INTO employee ...


Executes automatically.

Automatic execution

Trigger Flow
INSERT / UPDATE / DELETE
            ↓
       Trigger Fires
            ↓
      Execute Logic

:NEW
Value after change

BEFORE Trigger
Validation

Tracking

FOR EACH ROW
Runs once per row

Without FOR EACH ROW
Runs once per statement

Easy Memory Trick
BEFORE = Check
AFTER  = Track
