# Phone Number DataFrame

This notebook demonstrates how to create a Spark DataFrame containing phone numbers.

```python
from pyspark.sql.types import StructType, StructField, StringType, LongType

data = [
    ("Alice", "+1-555-123-4567"),
    ("Bob", "+1-555-987-6543"),
    ("Charlie", "+44-20-7946-0958"),
    ("Diana", "+91-98765-43210"),
    ("Eve", "+1-555-000-1111"),
]

schema = StructType([
    StructField("name", StringType(), nullable=False),
    StructField("phone_number", StringType(), nullable=False),
])

df = spark.createDataFrame(data, schema=schema)

display(df)
```

```python
from pyspark.sql.functions import lit, col, substring, concat

df1 = df.withColumn("masked_phone", concat(lit("******"),substring(col("phone_number"),7,4))) \
    .select(col("name"),col("masked_phone")).show()
```
